```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
k1 = float(input("Enter stiffness k1: "))
k2 = float(input("Enter stiffness k2: "))

print("k1 =", k1)
print("k2 =", k2)
```

```python
# Stiffness matrix of the simple 2-storey model
K = np.array([
    [k1 + k2, -k2],
    [-k2,      k2]
])

# Mass matrix is taken as identity for this simple model
M = np.eye(2)

print("\nStiffness matrix K:")
print(K)

print("\nMass matrix M:")
print(M)

# Solve Kx = lambda Mx
eigenvalues, eigenvectors = np.linalg.eigh(K)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors (columns):")
print(eigenvectors)

# Natural frequencies
frequencies = np.sqrt(eigenvalues)

print("\nNatural frequencies (in model units):")
print(frequencies)
```

```python
print("========== MATHEMATICAL RESULT ==========")

for i in range(2):
    print(f"\nMode {i+1}")
    print("Eigenvalue =", round(eigenvalues[i], 4))
    print("Eigenvector =", np.round(eigenvectors[:, i], 4))
    print("Frequency =", round(frequencies[i], 4))

print("\nEquation:")
print("K φ = λ M φ")
```

```python
# Floor positions
floors = np.array([0, 1, 2])

# Scale mode shapes so they are easy to see
scale = 0.7

plt.figure(figsize=(10, 5))

for i in range(2):
    mode = eigenvectors[:, i]

    # Add ground displacement = 0
    displacement = np.array([0, mode[0], mode[1]]) * scale

    plt.subplot(1, 2, i + 1)

    # Original building
    plt.plot([0, 0, 0], floors, 'k--', label="Original")

    # Vibrating building
    plt.plot(displacement, floors, 'o-', label=f"Mode {i+1}")

    plt.axvline(0, linestyle='--')
    plt.xlabel("Displacement")
    plt.ylabel("Floor")
    plt.title(f"Mode {i+1}")

    plt.yticks([0, 1, 2], ["Ground", "Floor 1", "Floor 2"])
    plt.grid(True)
    plt.legend()

plt.tight_layout()
plt.show()
```
